```python
import pandas as pd
import altair as alt
df=pd.read_csv("https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/building_inventory.csv")
df=df[['Agency Name', 'Year Constructed']].dropna()
df
```

```python
chart1=alt.Chart(df).mark_bar().encode(alt.X('count()', axis=alt.Axis(title='Number of Buildings')),alt.Y('Agency Name:N',sort ='-x')).properties(width=600,height=500,title='Number of Buildings by Illinois Agency')
alt.data_transformers.disable_max_rows()
chart1
```

```python
df_dropdown=alt.binding_select(options=sorted(df['Agency Name'].dropna().unique().tolist()), name='Agency: ')
selection = alt.selection_point(fields=['Agency Name'],bind=df_dropdown,value='University of Illinois')
chart2=alt.Chart(df).mark_bar().encode(alt.X('Year Constructed:Q',bin=alt.Bin(maxbins=30),axis=alt.Axis(title='Year Constructed', format='d')),alt.Y('count()', axis=alt.Axis(title='Number of Buildings'))).transform_filter(alt.datum['Year Constructed'] > 0).add_params(selection).transform_filter(selection).properties(width=600,height=350, title='Building Construction Years by Agency')
chart2
```

```python
chart1.save('chart1.json')
chart2.save('chart2.json')
```
